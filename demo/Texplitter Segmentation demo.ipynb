```python
from PIL import ImageFont, ImageDraw, Image

def get_text_size(text, font_size=12):
    # Path to the Inconsolata font installed on Fedora
    font_path = "/usr/share/fonts/google-inconsolata/Inconsolata-Regular.ttf"
    
    # Load the font
    font = ImageFont.truetype(font_path, font_size)
    
    # Create an image with a white background
    image = Image.new('RGB', (1000, 1000), color=(255, 255, 255))
    draw = ImageDraw.Draw(image)
    
    # Get the bounding box of the text
    bbox = draw.textbbox((0, 0), text, font=font)
    
    # Calculate the width and height
    width = bbox[2] - bbox[0]
    height = bbox[3] - bbox[1]
    
    return width, height

# Example usage
text = "This is a sample text to measure"
font_size = 12
width, height = get_text_size(text, font_size)
print(f"Width: {width}, Height: {height}")


```

```python
import json
import re

# Read the JSON file
with open('../src/texplitter/config/rules/default.json', 'r', encoding='utf-8') as file:
    patterns = json.load(file)

new_patterns = []
for p in patterns:
    text = p['description']
    
    # Step 1: Remove content before the first colon, including the colon itself
    text_after_colon = re.sub(r'^[^:]*:\s*', '', text)

    # Step 2: Replace \n followed by one or more spaces with a single space
    cleaned_text = re.sub(r'\n\s+', ' ', text_after_colon)

    p['description'] = cleaned_text
    new_patterns.append(p)

# Save the new patterns back to a JSON file with pretty formatting
with open('./regex_patterns_pretty.json', 'w', encoding='utf-8') as file:
    json.dump(new_patterns, file, ensure_ascii=False, indent=4)
```

```python
import openai
import json
import os

# Retrieve the OpenAI API key from the environment variable
openai_api_key = os.getenv('TEXPLITTER_OPENAI_CREDENTIAL')

if openai_api_key is None:
    raise ValueError("TEXPLITTER_OPENAI_CREDENTIAL environment variable is not set")

# Initialize the OpenAI client
client = openai.OpenAI(api_key=openai_api_key)

# Load the JSON file
with open('../src/texplitter/config/rules/default.json', 'r', encoding='utf-8') as file:
    patterns = json.load(file)

# Extract the regex patterns for analysis
regex_patterns = [p['regex'] for p in patterns]

# Create the prompt
prompt = f"""
I have a list of regular expressions that are evaluated in sequence. I want to ensure that none of these regular expressions will interfere with the intended parsing logic because of the order in which they are evaluated. Specifically, a regular expression might interfere if an earlier regex matches part of the text that a later one is supposed to match.

Here is the list of regular expressions in verbose format:
"""

# Append each pattern in verbose format
for i, pattern in enumerate(patterns, start=1):
    prompt += f"\n{i}. {pattern['regex']}"

prompt += "\n\nPlease analyze these regular expressions and identify any potential issues with their order that might affect their application. For each pair, indicate if the earlier one would interfere with the later one and suggest a correct order if needed."

# Call the GPT-4 model using the updated API
response = client.chat.completions.create(
    model="gpt-4",
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": prompt}
    ]
)

# Print the analysis
analysis = response.choices[0].message.content.strip()
print(analysis)
```
